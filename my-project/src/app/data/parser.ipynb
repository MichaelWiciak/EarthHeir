```python
import json

# Load the JSON data from the file
with open('phone.json', 'r') as file:
    data = json.load(file)

# Clean and update the data
cleaned_data = {
    country: value if value.startswith('+') else f'+{value}'
    for country, value in data.items()
    if value.strip() != ''
}

# print some of the cleaned data
for country, value in list(cleaned_data.items())[:5]:  # Print first
    print(f"{country}: {value}")
```

```python
# Load country names
with open('names.json', 'r') as f:
    name_data = json.load(f)

# Create new mapping: Country Name → Phone Code
country_to_code = {
    name_data[code]: cleaned_data[code]
    for code in cleaned_data
    if code in name_data
}

# print some of the country to code mapping
for country, code in list(country_to_code.items())[:5]:  # Print first
    print(f"{country}: {code}")

# print out the code of United Kingdom
print(f"United Kingdom: {country_to_code.get('United Kingdom', 'Not found')}")


```

```python
country_to_code = dict(sorted(country_to_code.items()))

# Print the sorted country to code mapping
for country, code in list(country_to_code.items())[:5]:
    print(f"{country}: {code}")
```

```python
# Save the updated data as phoneNodes.json
with open('phoneNodes.json', 'w') as file:
    json.dump(country_to_code, file, indent=4)
```
